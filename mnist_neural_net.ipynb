```python
train_zero = torch.stack([tensor(Image.open(o)) for o in zero]).float()/255
train_one = torch.stack([tensor(Image.open(o)) for o in one]).float()/255
train_two = torch.stack([tensor(Image.open(o)) for o in two]).float()/255
train_three = torch.stack([tensor(Image.open(o)) for o in three]).float()/255
train_four = torch.stack([tensor(Image.open(o)) for o in four]).float()/255
train_five = torch.stack([tensor(Image.open(o)) for o in five]).float()/255
train_six = torch.stack([tensor(Image.open(o)) for o in six]).float()/255
train_seven = torch.stack([tensor(Image.open(o)) for o in seven]).float()/255
train_eight = torch.stack([tensor(Image.open(o)) for o in eight]).float()/255
train_nine = torch.stack([tensor(Image.open(o)) for o in nine]).float()/255
```

```python
valid_zero= (path/'testing'/'0').ls().sorted()
valid_one= (path/'testing'/'1').ls().sorted()
valid_two= (path/'testing'/'2').ls().sorted()
valid_three= (path/'testing'/'3').ls().sorted()
valid_four= (path/'testing'/'4').ls().sorted()
valid_five= (path/'testing'/'5').ls().sorted()
valid_six= (path/'testing'/'6').ls().sorted()
valid_seven= (path/'testing'/'7').ls().sorted()
valid_eight= (path/'testing'/'8').ls().sorted()
valid_nine= (path/'testing'/'9').ls().sorted()
```

```python
test_zero = torch.stack([tensor(Image.open(o)) for o in valid_zero]).float()/255
test_one = torch.stack([tensor(Image.open(o)) for o in valid_one]).float()/255
test_two = torch.stack([tensor(Image.open(o)) for o in valid_two]).float()/255
test_three = torch.stack([tensor(Image.open(o)) for o in valid_three]).float()/255
test_four = torch.stack([tensor(Image.open(o)) for o in valid_four]).float()/255
test_five = torch.stack([tensor(Image.open(o)) for o in valid_five]).float()/255
test_six = torch.stack([tensor(Image.open(o)) for o in valid_six]).float()/255
test_seven = torch.stack([tensor(Image.open(o)) for o in valid_seven]).float()/255
test_eight = torch.stack([tensor(Image.open(o)) for o in valid_eight]).float()/255
test_nine = torch.stack([tensor(Image.open(o)) for o in valid_nine]).float()/255
```

```python
train_x= torch.cat([train_zero, train_one, train_two, train_three, train_four, train_five, train_six, train_seven, train_eight,train_nine]).view(-1, 28*28)
train_y = tensor([0]*len(zero) + [1]*len(one) + [2]*len(two) + [3]*len(three) + [4]*len(four) + [5]*len(five) +[6]*len(six) + [7]*len(seven) + [8]*len(eight) + [9]*len(nine))#.unsqueeze(1)
train_x.shape, train_y.shape
```

```python
valid_x= torch.cat([test_zero, test_one, test_two, test_three, test_four, test_five, test_six, test_seven, test_eight,test_nine]).view(-1, 28*28)
valid_y = tensor([0]*len(valid_zero) + [1]*len(valid_one) + [2]*len(valid_two) + [3]*len(valid_three) + [4]*len(valid_four) + [5]*len(valid_five) +[6]*len(valid_six) + [7]*len(valid_seven) + [8]*len(valid_eight) + [9]*len(valid_nine))#.unsqueeze(1)
valid_x.shape, valid_y.shape
```

```python
dset= list(zip(train_x, train_y))
valid_dset=list(zip(valid_x, valid_y))
```

```python
dl= DataLoader(dset, batch_size=5000)
valid_dl= DataLoader(valid_dset, batch_size=1000)
```

```python
dls= DataLoaders(dl, valid_dl)
```

```python
def init_param(size, std=1.0): return (torch.randn(size)*std).requires_grad_()
```

```python
def mnist_loss(prediction, target):
  criterion = nn.CrossEntropyLoss()
  return criterion(prediction, target)
```

```python
def batch_accuracy(xb, yb):
  preds=xb.argmax(dim=1)
  correct = (preds==yb)
  return correct.float().mean()
```

```python
simple_net = nn.Sequential(
    nn.Linear(28*28, 300),
    nn.ReLU(),
    nn.Linear(300, 100),
    nn.ReLU(),
    nn.Linear(100, 10),

)
```

```python
learn = Learner(dls, simple_net, opt_func=SGD,
                loss_func=mnist_loss,metrics=batch_accuracy, )
```
