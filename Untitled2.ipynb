```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
order_details = pd.read_csv("order_details.csv")
order = pd.read_csv("orders.csv")
types_of_pizza = pd.read_csv("pizza_types.csv", encoding="latin1")
pizza = pd.read_csv("pizzas.csv")
data_1 = pd.merge(order_details, order)
data_2 = pd.merge(types_of_pizza, pizza)
data = pd.merge(data_1, data_2)
data
#order_details
```
